```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Filas anonimizadas (LGPD)
fila_prioridade = []
fila_normal = []

# Componentes da Interface
output = widgets.Output()
id_input = widgets.Text(placeholder='Digite o ID/Senha', description='ID:')
btn_normal = widgets.Button(description='Adicionar Normal', button_style='info')
btn_prioridade = widgets.Button(description='Adicionar Prioridade', button_style='danger')
btn_proximo = widgets.Button(description='CHAMAR PRÓXIMO', button_style='success')
label_status = widgets.Label(value="Fila vazia")

def atualizar_view(chamada="Aguardando Chamada..."):
    with output:
        clear_output()
        total = len(fila_prioridade) + len(fila_normal)
        print(f"--- STATUS DA FILA ---")
        print(f"Aguardando: {total} | Prioridades: {len(fila_prioridade)}")
        print(f"\nPAINEL: {chamada}")

def add_normal(b):
    if id_input.value:
        fila_normal.append(id_input.value)
        id_input.value = ''
        atualizar_view()

def add_prioridade(b):
    if id_input.value:
        fila_prioridade.append(id_input.value)
        id_input.value = ''
        atualizar_view()

def chamar(b):
    if fila_prioridade:
        proximo = fila_prioridade.pop(0)
        atualizar_view(f"PRIORIDADE: {proximo}")
    elif fila_normal:
        proximo = fila_normal.pop(0)
        atualizar_view(f"NORMAL: {proximo}")
    else:
        atualizar_view("FILA VAZIA")

# Vincular funções aos botões
btn_normal.on_click(add_normal)
btn_prioridade.on_click(add_prioridade)
btn_proximo.on_click(chamar)

# Exibir Interface
display(id_input, widgets.HBox([btn_normal, btn_prioridade]), btn_proximo, output)
atualizar_view()
```
