```python
import random

def jogar():
    # 1. Configurações Iniciais
    numeros_disponiveis = [1, 2, 3, 4, 5, 6, 7, 8]
    tamanho_codigo = 5
    tentativas_maximas = 5


    # Gera o código secreto (sem repetições para facilitar a lógica)
    codigo_secreto = random.sample(numeros_disponiveis, tamanho_codigo)

    print("="*40)
    print("🎯 BEM-VINDO AO ENIGMA NUMÉRICO!")
    print(f"Descubra o código de {tamanho_codigo} números (de 1 a 8).")
    print("Dicas: ✅ = Posição Correta | ○ = Número certo, posição errada")
    print("="*40)

    # 2. Loop de Tentativas
    for t in range(1, tentativas_maximas + 1):
        try:
            entrada = input(f"\nTentativa {t}/{tentativas_maximas} - Digite 5 números: ").split()
            chute = [int(n) for n in entrada]

            if len(chute) != tamanho_codigo:
                print(f"⚠️ Digite exatamente {tamanho_codigo} números!")
                continue
        except ValueError:
            print("⚠️ Digite apenas números inteiros!")
            continue

        # 3. Verificação (Lógica Principal)
        posicao_correta = 0
        numero_correto_posicao_errada = 0

        for i in range(tamanho_codigo):
            if chute[i] == codigo_secreto[i]:
                posicao_correta += 1
            elif chute[i] in codigo_secreto:
                numero_correto_posicao_errada += 1

        # 4. Resultados
        if posicao_correta == tamanho_codigo:
            print(f"\n🎉 PARABÉNS! Você venceu em {t} tentativas!")
            print(f"Código: {codigo_secreto}")
            return

        print(f"Dicas: {'✅' * posicao_correta}{'○' * numero_correto_posicao_errada}")

    print(f"\n😢 Que pena! O código era: {codigo_secreto}")

if __name__ == "__main__":
    jogar()
```
